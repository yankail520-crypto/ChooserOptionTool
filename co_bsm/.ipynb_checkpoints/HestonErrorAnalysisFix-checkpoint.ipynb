### 误差分析

本节比较 Black-Scholes-Merton 模型（BSM，常数波动率）与 Heston 随机波动率模型对同一份 chooser 期权的定价。chooser 为场外产品，基准价格由与普通期权市场价格相一致的 Heston 模型给出：以 2018–2024 年 84 个月度期权链快照（每月首个交易日）逐一校准 Heston 参数，拟合的隐含波动率 RMSE 均值为 1.10 个波动率点；快照之间的交易日沿用最近快照的 $\kappa,\theta,\xi,\rho$，$v_0$ 按 VIX 的变化缩放。样本为 2018-01-02 至 2024-12-31 共 1761 个交易日，每日对两份合约定价：$K=S_t$（固定 moneyness）与 $K=150$，$T_1=0.5$，$T_2=1$。Heston 价格采用 QE 格式的蒙特卡罗（每日 10 万条路径，$[0,T_1]$ 与 $[T_1,T_2]$ 各 32 步），并与特征函数积分得到的半解析解逐日对照。BSM 的波动率取两种设定：$\sigma_{\text{hist}}$ 为 20 日历史波动率，$\sigma_{\text{impl}}$ 为 Heston 在期限 $T_2$、行权价等于远期价格处的隐含波动率。

记 $\tau=T_2-T_1$，$K'=Ke^{-(r-q)\tau}$。在 $r,q$ 为确定性常数的条件下，chooser 的分解公式与模型无关，对任一模型 $M$ 有

$$
V^{M}=C^{M}(K,T_2)+e^{-q\tau}P^{M}(K',T_1).
$$

因此 BSM 与 Heston 的价差等于两条腿价差之和：

$$
\Delta V=V^{\text{BSM}}-V^{\text{H}}=\big[C^{\text{BSM}}-C^{\text{H}}\big](K,T_2)+e^{-q\tau}\big[P^{\text{BSM}}-P^{\text{H}}\big](K',T_1).
$$

记 $\sigma_C$、$\sigma_P$ 为 Heston 两条腿价格对应的 BSM 隐含波动率，即 $C^{\text{BSM}}(K,T_2;\sigma_C)=C^{\text{H}}(K,T_2)$，$P^{\text{BSM}}(K',T_1;\sigma_P)=P^{\text{H}}(K',T_1)$。由中值定理，

$$
\Delta V=\nu_C\,(\sigma-\sigma_C)+e^{-q\tau}\nu_P\,(\sigma-\sigma_P),
$$

其中 $\nu_C,\nu_P$ 为两条腿的 vega，$\sigma$ 为 BSM 采用的常数波动率。两条腿位于波动率曲面的不同位置，Call 腿在 $(K,T_2)$，Put 腿在 $(K',T_1)$，Heston 在两处给出不同的隐含波动率，BSM 对两处使用同一个 $\sigma$，价差由此产生。

对 $\sigma_{\text{hist}}$ 版本，总误差进一步分解为波动率估计误差与曲面形状误差：

$$
V^{\text{BSM}}(\sigma_{\text{hist}})-V^{\text{H}}
=\underbrace{V^{\text{BSM}}(\sigma_{\text{hist}})-V^{\text{BSM}}(\sigma_{\text{impl}})}_{\approx\,(\nu_C+e^{-q\tau}\nu_P)(\sigma_{\text{hist}}-\sigma_{\text{impl}})}
+\underbrace{V^{\text{BSM}}(\sigma_{\text{impl}})-V^{\text{H}}}_{\text{曲面形状误差}}.
$$

第一项反映历史波动率与市场隐含波动率的差异，第二项在两个模型的波动率水平对齐之后保留下来，反映常数波动率假设对偏斜与期限结构的忽略。

误差指标以 $e_t=V^{\text{BSM}}_t-V^{\text{H}}_t$ 为基础（$e_t>0$ 表示 BSM 高估），包括

$$
\text{MAE}=\frac1n\sum_t|e_t|,\quad
\text{RMSE}=\sqrt{\frac1n\sum_t e_t^2},\quad
\text{Bias}=\frac1n\sum_t e_t,\quad
\text{MAPE}=\frac{100}{n}\sum_t\frac{|e_t|}{V^{\text{H}}_t}.
$$

每项指标分别在全样本与两个子样本上计算。可靠子样本剔除校准参数触及搜索边界或拟合 RMSE 超过 2 个波动率点的快照（9 个快照，占交易日 10.7%），以及 $v_0$ 触及上界的日期；第二个子样本在此基础上再剔除 120–450 天合约的拟合 RMSE 超过 1.5 个波动率点的快照。

基准价格含三类不确定性：蒙特卡罗的统计误差，由标准误 $\text{se}=s/\sqrt{N}$ 度量；时间离散的系统偏差，由蒙特卡罗与半解析解之差的均值度量；Heston 对市场期权的校准误差，折算为价格约为 $(\nu_C+e^{-q\tau}\nu_P)\times\text{RMSE}_{\text{iv}}$，其中 $\text{RMSE}_{\text{iv}}$ 取两条腿所在期限（120–450 天）的合约。BSM 与 Heston 的价差以这三项为尺度进行解读，超出该尺度的部分归因于模型差异。

```python
import sys
from pathlib import Path

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "co_bsm").is_dir())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import cm
from scipy.stats import norm, rankdata

from co_bsm import calibrate as cal
from co_bsm.black_scholes import BSMParams, bsm_call, bsm_put
from co_bsm.chooser import ChooserParams, chooser_price
from co_bsm.implied_vol import implied_vol
from co_bsm.pricing import HestonPricer
from co_data_center.data_center import DataCenter
from common.paths import get_outputs_path

plt.rcParams["font.sans-serif"] = ["SimHei", "Microsoft YaHei", "PingFang SC",
                                   "Noto Sans CJK SC", "WenQuanYi Zen Hei", "DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False
pd.set_option("display.float_format", "{:.4f}".format)
pd.set_option("display.width", 200)

out = get_outputs_path()
prices = pd.read_parquet(out / "chooser_daily_prices.parquet")
params = pd.read_parquet(out / "heston_daily_params.parquet")[["date", "snapshot_date", "rmse_iv_snapshot"]]
feat = DataCenter().get("feature_daily").df
feat = feat.assign(date=pd.to_datetime(feat["date"]))[["date", "vix", "sentiment_score", "news_count"]]

df = prices.merge(params, on="date", how="left").merge(feat, on="date", how="left")
df["reliable"] = ~df["low_conf"] & (df["v0"] < 0.999)      # v0 被截断在上界的日期也算不可靠
BENCH = "heston_mc"

print(f"{df['date'].nunique()} 个交易日，{len(df)} 行；可靠的行占 {df['reliable'].mean():.1%}")
print("MC 列是否完整：", df[BENCH].notna().all())
```

### 1.1 统计误差

```python
mc = df.dropna(subset=[BENCH, "heston_mc_se"]).copy()
mc["z"] = (mc[BENCH] - mc["heston_exact"]) / mc["heston_mc_se"]
mc["rel_se_pct"] = mc["heston_mc_se"] / mc["heston_exact"] * 100

stat = mc.groupby("contract").agg(
    天数=("z", "size"),
    标准误中位数=("heston_mc_se", "median"),
    相对标准误中位数_pct=("rel_se_pct", "median"),
    z均值=("z", "mean"),
    z标准差=("z", "std"),
    超过2的占比=("z", lambda s: (s.abs() > 2).mean()),
    超过3的占比=("z", lambda s: (s.abs() > 3).mean()),
)
print("纯抽样噪声下：z 均值 0，标准差 1，超过 2 的占比 4.6%，超过 3 的占比 0.27%")
display(stat)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2), sharey=True)
x = np.linspace(-4, 4, 200)
for ax, (name, g) in zip(axes, mc.groupby("contract")):
    ax.hist(g["z"], bins=40, range=(-4, 4), density=True, alpha=0.6, label="实际")
    ax.plot(x, norm.pdf(x), "k--", label="N(0, 1)")
    ax.set(title=f"{name}：(MC - 精确值) / 标准误", xlabel="z")
    ax.legend()
plt.tight_layout()
plt.show()
```

**统计误差.** 每日 10 万条路径下，蒙特卡罗价格的标准误中位数为 0.069（$K=S_t$）与 0.071（$K=150$），分别占价格的 0.29% 与 0.21%。$z_t=(\hat V_t-V_t^{\text{exact}})/\text{se}_t$ 的标准差为 0.98 与 1.01，绝对值超过 2 的占比均为 4.6%，超过 3 的占比为 0.17% 与 0.34%，与标准正态分布的 4.6% 和 0.27% 吻合，所报告的标准误如实反映了抽样误差。$z_t$ 的均值为 $-0.066$ 与 $-0.054$，对应下一节的平均偏差。

### 1.2 离散误差

```python
mc["bias"] = mc[BENCH] - mc["heston_exact"]
mc["regime"] = np.select([mc["xi"] >= 3.9, mc["kappa"] >= 24.9], ["ξ 撞界", "κ 撞界"], default="其他")


def bias_table(keys):
    rows, index = [], []
    for k, g in mc.groupby(keys):
        n = len(g)
        mean = g["bias"].mean()
        se = np.sqrt((g["heston_mc_se"] ** 2).sum()) / n       # 各天的 MC 相互独立
        rows.append({"天数": n, "平均偏差": mean, "均值标准误": se, "偏差/标准误": mean / se,
                     "平均相对偏差_pct": (g["bias"] / g["heston_exact"]).mean() * 100})
        index.append(k)
    if isinstance(keys, list):
        return pd.DataFrame(rows, index=pd.MultiIndex.from_tuples(index, names=keys))
    return pd.DataFrame(rows, index=pd.Index(index, name=keys))


print("按合约：")
display(bias_table("contract"))
print("按参数状态：")
display(bias_table(["contract", "regime"]))
print("按可靠性：")
display(bias_table(["contract", "reliable"]))
```

**离散误差.** 每日的蒙特卡罗价格与精确值之差取平均后，抽样噪声互相抵消，剩下的是时间离散带来的系统偏差。平均偏差为 $-0.005$（$K=S_t$）与 $-0.003$（$K=150$），分别是均值标准误的 3.1 倍与 1.5 倍，占价格的 0.018% 与 0.011%。参数触及边界的日期偏差更大：$K=S_t$ 合约在 $\kappa$ 触界的 19 天为 $-0.042$（0.15%），在 $\xi$ 触界的 147 天为 $-0.014$；$K=150$ 合约对应为 $+0.006$（不显著）与 $-0.012$。可靠日期的平均偏差为 $-0.004$，不可靠日期为 $-0.012$（$K=S_t$）。步长检验在四组校准得到的真实参数上进行：每日定价采用的每段 32 步下，QE 的偏差在 $-0.04$ 至 $+0.04$ 之间，不超过 1.0 个标准误；Full Truncation 的偏差为 $+0.21$ 至 $+1.06$，为 4.9 至 12.8 个标准误，因此基准采用 QE。

### 1.3 校准误差

```python
PN = ["v0", "kappa", "theta", "xi", "rho"]
hp = pd.read_parquet(out / "heston_params.parquet")
hp = hp[hp["status"] == "ok"].set_index("date")
chain, featc = cal.load_inputs()


def iv_surface(p, ks, Ts):
    """Heston 隐含波动率，只取决于 ln(K/F) 和期限，与 S、r、q 无关。返回 (len(Ts), len(ks))。"""
    S = 100.0
    K = S * np.exp(ks)
    otm = ks >= 0
    res = np.empty((len(Ts), len(ks)))
    for i, T in enumerate(Ts):
        c = HestonPricer(S, 0.0, 0.0, np.full(len(K), T), K).call_prices(**p)
        res[i] = implied_vol(np.where(otm, c, c - (S - K)), S, K, 0.0, T, 0.0, otm)
    return res


def snapshot_options(date):
    S, r, q = cal.market_inputs(featc, date)
    opts, _ = cal.select_options(chain[chain["date"] == date], S, r, q)
    opts["k"] = np.log(opts["K"] / (S * np.exp((r - q) * opts["T"])))
    return opts
```

```python
def plot_snapshot(date, n_curves=6):
    date = pd.Timestamp(date)
    row = hp.loc[date]
    p = {k: float(row[k]) for k in PN}
    opts = snapshot_options(date)

    Ts = np.sort(opts["T"].unique())
    pick = Ts[np.unique(np.linspace(0, len(Ts) - 1, n_curves).round().astype(int))]
    ks = np.linspace(-0.35, 0.35, 71)
    fig, axes = plt.subplots(2, 3, figsize=(14, 8), sharex=True)
    for ax, T in zip(axes.flat, pick):
        m = opts[np.isclose(opts["T"], T)]
        ax.scatter(m["k"], m["iv"] * 100, s=14, label="市场（中间价反算）")
        ax.plot(ks, iv_surface(p, ks, np.array([T]))[0] * 100, "r-", lw=1.6, label="Heston")
        ax.set(title=f"期限 {T * 365:.0f} 天（{len(m)} 个合约）", ylabel="隐含波动率 (%)")
        ax.grid(alpha=0.3)
    axes[0, 0].legend()
    for ax in axes[-1]:
        ax.set_xlabel("ln(K / F)")
    fig.suptitle(f"{date.date()}：Heston 与市场隐含波动率（拟合 RMSE {row['rmse_iv']:.2f} 个波动率点，"
                 f"κ={p['kappa']:.1f}，ξ={p['xi']:.2f}，ρ={p['rho']:.2f}）")
    plt.tight_layout()
    plt.show()


for dt in ["2021-12-01", "2020-04-01", "2024-06-03"]:
    plot_snapshot(dt)
```

```python
def fit_errors(date):
    """某个快照上各合约的拟合误差，单位为波动率点（模型 - 市场）"""
    p = {k: float(hp.loc[date, k]) for k in PN}
    opts = snapshot_options(date)
    model = np.empty(len(opts))
    for T in opts["T"].unique():
        m = np.asarray(np.isclose(opts["T"], T))
        model[m] = iv_surface(p, opts.loc[m, "k"].to_numpy(), np.array([T]))[0]
    return pd.DataFrame({"date": date, "days": opts["T"].to_numpy() * 365, "k": opts["k"].to_numpy(),
                         "err": (model - opts["iv"].to_numpy()) * 100})


def rmse(s):
    return np.sqrt((s ** 2).mean())


errs = pd.concat([fit_errors(dt) for dt in hp.index], ignore_index=True)

# 重算的 RMSE 应与校准时记录的一致
snap_rmse = errs.groupby("date")["err"].agg(rmse)
print("与 heston_params 记录的 rmse_iv 的最大差：", float((snap_rmse - hp["rmse_iv"]).abs().max()))

errs["期限"] = pd.cut(errs["days"], [0, 45, 120, 240, 450, 800],
                    labels=["<45 天", "45–120 天", "120–240 天", "240–450 天", ">450 天"])
errs["位置"] = pd.cut(errs["k"], [-1, -0.1, 0.1, 1], labels=["左翼 (k<-0.1)", "平值附近", "右翼 (k>0.1)"])

display(errs.groupby("期限", observed=True)["err"].agg(
    合约数="size", 平均误差="mean", RMSE=rmse, 绝对误差中位数=lambda s: s.abs().median()))
print("平均误差（模型 - 市场，波动率点），按期限和行权价位置：")
display(errs.pivot_table(index="期限", columns="位置", values="err", aggfunc="mean", observed=True))

# 两条腿所在期限（120-450 天）的拟合误差
rel = errs[errs["days"].between(120, 450)].groupby("date")["err"].agg(rmse)
print(f"各快照整体 RMSE 均值 {hp['rmse_iv'].mean():.2f}；120–450 天合约：均值 {rel.mean():.2f}，"
      f"中位数 {rel.median():.2f}，最大 {rel.max():.2f}")

df["rmse_rel"] = df["snapshot_date"].map(rel)
df["fit_ok"] = df["reliable"] & (df["rmse_rel"] <= 1.5)
```

```python
# 撞界快照在两条腿所在期限上是否拟合更差
bound = ((hp["kappa"] >= 24.9) | (hp["xi"] >= 3.9)).reindex(rel.index)
display(pd.DataFrame({"撞界": rel[bound].describe(), "未撞界": rel[~bound].describe()}).round(2))

weak = rel[rel > 1.5]
print("120–450 天 RMSE 超过 1.5 的快照：", {str(k.date()): round(v, 2) for k, v in weak.items()})
print("其中已标为低可信度：", [str(k.date()) for k in weak.index if df.loc[df["snapshot_date"] == k, "low_conf"].any()])

# 用 120-450 天的 RMSE 折算校准不确定性，与 BSM 和 Heston 的差距比较
d = df.dropna(subset=["sigma_impl", "rmse_rel"]).copy()


def chooser_at(sig):
    return chooser_price(ChooserParams(S=d["S"].to_numpy(), K=d["K"].to_numpy(), r=d["r"].to_numpy(),
                                       T1=d["T1"].to_numpy(), T2=d["T2"].to_numpy(),
                                       sigma=sig, q=d["q"].to_numpy()))


s = d["sigma_impl"].to_numpy()
d["vega_per_pt"] = chooser_at(s + 0.005) - chooser_at(s - 0.005)      # sigma 变动 1 个点的价格变化
d["calib_unc"] = d["vega_per_pt"] * d["rmse_rel"]
d["gap_impl"] = (d["bsm_impl"] - d[BENCH]).abs()
d["gap_gt_unc"] = d["gap_impl"] > d["calib_unc"]

display(d.groupby("contract").agg(
    相关期限RMSE_点=("rmse_rel", "median"),
    chooser_vega每点=("vega_per_pt", "median"),
    校准不确定性_中位数=("calib_unc", "median"),
    MC标准误_中位数=("heston_mc_se", "median"),
    BSM与Heston绝对差_中位数=("gap_impl", "median"),
    差距大于校准不确定性的占比=("gap_gt_unc", "mean")))
```

**校准误差.** 84 个快照共 19940 个参与校准的合约，Heston 拟合隐含波动率的 RMSE 为 1.10 个波动率点。按期限分层，RMSE 呈两头高、中间低：期限短于 45 天为 1.41（占合约数的 42%），45–120 天为 1.16，120–240 天为 0.80，240–450 天为 0.75，长于 450 天为 1.18。chooser 的两条腿分别位于约 182 天（$T_1=0.5$）与 365 天（$T_2=1$），落在拟合最好的区间；各快照在 120–450 天合约上的 RMSE 均值为 0.72、中位数为 0.60、最大值为 2.27。拟合误差有系统性的一面：在期限不短于 240 天的合约上，模型相对市场的平均误差在左翼为 $-0.27$ 至 $-0.40$、右翼为 $+0.25$ 至 $+0.51$ 个波动率点，校准得到的偏斜在长期限上比市场平缓。在拟合偏弱的三个快照（2021-12-01、2020-04-01、2024-06-03）上，较大的偏差同样出现在最短与最长的期限：2021-12-01 的 16 天期限上，市场左翼的隐含波动率高于 $\xi$ 触及上界的曲线；2020-04-01 的 16 天期限上，平值附近的市场隐含波动率在 60% 至 91% 之间分散；2024-06-03 的 592 天期限上，市场左翼高于曲线 2–3 个点、右翼低于曲线约 1 个点。

以 120–450 天合约的 RMSE 为准，触及搜索边界的 8 个快照的均值为 0.84（中位数 0.65），其余 76 个为 0.70（0.59），差别不大。该 RMSE 超过 1.5 的快照有 6 个，全部在 2020 年 4 月至 9 月，其中只有 2020-04-01 被标为低可信度。按这一 RMSE 折算，校准不确定性的中位数为 0.58（$K=S_t$）与 0.39（$K=150$），分别是蒙特卡罗标准误的 8.5 倍与 5.5 倍，是基准最主要的不确定性来源。

### 2 BSM 相对基准的表现

```python
def metrics(g, col, bench=BENCH):
    e = g[col] - g[bench]
    pct = e / g[bench] * 100
    return pd.Series({"天数": len(g), "MAE": e.abs().mean(), "RMSE": np.sqrt((e ** 2).mean()),
                      "平均偏差": e.mean(), "偏差中位数": e.median(),
                      "MAPE_pct": pct.abs().mean(), "平均相对偏差_pct": pct.mean(),
                      "P95绝对误差": e.abs().quantile(0.95)})


versions = [("bsm_hist", "σ = 历史波动率"), ("bsm_impl", "σ = Heston 隐含")]
rows = []
for contract, g in df.groupby("contract"):
    subsets = [("全样本", g), ("仅可靠日期", g[g["reliable"]]), ("再剔除相关期限拟合偏弱的快照", g[g["fit_ok"]])]
    for subset, h in subsets:
        for col, label in versions:
            rows.append(metrics(h.dropna(subset=[col, BENCH]), col).rename((contract, subset, label)))
baseline = pd.DataFrame(rows)
baseline.index.names = ["合约", "样本", "BSM 版本"]
display(baseline)
baseline.to_csv(out / "bsm_baseline_metrics.csv", encoding="utf-8-sig")
print("偏差 = BSM - Heston，为正表示 BSM 高估；已保存 bsm_baseline_metrics.csv")

# 基准换成半解析精确值，结果应几乎不变
main = [np.sqrt(((g["bsm_hist"] - g[BENCH]) ** 2).mean()) for _, g in df.groupby("contract")]
alt = [np.sqrt(((g["bsm_hist"] - g["heston_exact"]) ** 2).mean()) for _, g in df.groupby("contract")]
print("RMSE（σ=历史波动率）：以 MC 为基准", np.round(main, 4), "；以精确值为基准", np.round(alt, 4))
```

```python
# BSM 价格是确定的，它与 MC 价格之差的标准误就是 MC 的标准误
rows = []
for contract, g in df.dropna(subset=["bsm_hist", "bsm_impl", BENCH]).groupby("contract"):
    for col, label in versions:
        t = (g[col] - g[BENCH]).abs() / g["heston_mc_se"]
        rows.append({"合约": contract, "BSM 版本": label, "差距/标准误_中位数": t.median(),
                     "差距超过2倍标准误的占比": (t > 2).mean(), "差距超过3倍标准误的占比": (t > 3).mean()})
display(pd.DataFrame(rows).set_index(["合约", "BSM 版本"]))
```

### 2.1 误差分解

```python
# 总误差 = 波动率估计误差 + 曲面形状误差
# bsm_hist - heston = (bsm_hist - bsm_impl) + (bsm_impl - heston)
rows = []
for contract, g in df.dropna(subset=["bsm_hist", "bsm_impl", BENCH]).groupby("contract"):
    for subset, h in [("全样本", g), ("仅可靠日期", g[g["reliable"]])]:
        parts = {"总误差": h["bsm_hist"] - h[BENCH],
                 "波动率估计误差": h["bsm_hist"] - h["bsm_impl"],
                 "曲面形状误差": h["bsm_impl"] - h[BENCH]}
        for name, e in parts.items():
            rows.append({"合约": contract, "样本": subset, "分量": name,
                         "平均": e.mean(), "RMSE": np.sqrt((e ** 2).mean())})
display(pd.DataFrame(rows).set_index(["合约", "样本", "分量"]))
```

```python
def bsm_legs(g, sigma):
    """BSM 下 chooser 的两条腿：Call(K, T2) 与 e^{-q tau} Put(K', T1)"""
    S, K, T1, T2 = (g[c].to_numpy() for c in ("S", "K", "T1", "T2"))
    r, q = g["r"].to_numpy(), g["q"].to_numpy()
    tau = T2 - T1
    Kp = K * np.exp(-(r - q) * tau)
    call = bsm_call(BSMParams(S, K, r, T2, sigma, q))
    put = np.exp(-q * tau) * bsm_put(BSMParams(S, Kp, r, T1, sigma, q))
    return call, put


rows = []
for contract, g in df.dropna(subset=["sigma_hist", "sigma_impl"]).groupby("contract"):
    for subset, h in [("全样本", g), ("仅可靠日期", g[g["reliable"]])]:
        for col, label in [("sigma_hist", "σ = 历史波动率"), ("sigma_impl", "σ = Heston 隐含")]:
            sigma = h[col].to_numpy()
            ok = np.isfinite(sigma) & (sigma > 0)
            hh = h[ok]
            call, put = bsm_legs(hh, sigma[ok])
            for leg, bsm_v, hes_v in [("Call 腿", call, hh["heston_call_leg"].to_numpy()),
                                      ("Put 腿", put, hh["heston_put_leg"].to_numpy())]:
                e = bsm_v - hes_v
                rows.append({"合约": contract, "样本": subset, "BSM 版本": label, "腿": leg,
                             "平均偏差": e.mean(), "RMSE": np.sqrt((e ** 2).mean())})
display(pd.DataFrame(rows).set_index(["合约", "样本", "BSM 版本", "腿"]))     # Heston 两条腿取半解析精确值
```

**BSM 的表现.** 以 20 日历史波动率为输入，BSM 相对基准的 RMSE 为 9.44（$K=S_t$）与 7.33（$K=150$），MAPE 为 27.2% 与 13.1%，平均偏差为 $-1.21$ 与 $-0.55$，中位数偏差为 $-3.28$ 与 $-0.94$，均值高于中位数是因为高波动时期的高估拉高了均值；仅用可靠日期时 RMSE 为 7.47 与 5.09。以 Heston 在 $T_2$ 期限、远期平值处的隐含波动率为输入，RMSE 降至 0.50 与 1.34，MAPE 为 1.85% 与 3.24%，平均偏差为 $+0.30$ 与 $+0.41$，仅用可靠日期时 RMSE 为 0.47 与 1.34。总误差以波动率估计误差为主（RMSE 9.54 与 7.11），曲面形状误差的 RMSE 为 0.50 与 1.34。$K=S_t$ 合约的形状误差集中在 Put 腿（平均 $+0.30$），Call 腿接近 0，这是构造所致，因为隐含波动率取自 Call 腿所在位置；$K=150$ 合约两条腿的偏差均为正（$+0.18$ 与 $+0.22$）。基准换成半解析精确值时，RMSE 的变化在小数点后第四位。以蒙特卡罗标准误计，两个版本与基准的差距中位数分别为 78 倍与 32 倍（历史波动率）、5.7 倍与 13 倍（Heston 隐含），都远超抽样噪声。

```python
# 逐日的两条腿隐含波动率，单位为波动率点
dd = df.dropna(subset=["sigma_impl", "sigma_hist"]).copy()
r, q = dd["r"].to_numpy(), dd["q"].to_numpy()
tau = (dd["T2"] - dd["T1"]).to_numpy()
Kp = dd["K"].to_numpy() * np.exp(-(r - q) * tau)
put_price = dd["heston_put_leg"].to_numpy() / np.exp(-q * tau)      # 还原为 Put(K', T1) 本身

dd["sigma_C"] = implied_vol(dd["heston_call_leg"].to_numpy(), dd["S"].to_numpy(), dd["K"].to_numpy(),
                            r, dd["T2"].to_numpy(), q, True)
dd["sigma_P"] = implied_vol(put_price, dd["S"].to_numpy(), Kp, r, dd["T1"].to_numpy(), q, False)
dd["C-impl"] = (dd["sigma_C"] - dd["sigma_impl"]) * 100
dd["P-C"] = (dd["sigma_P"] - dd["sigma_C"]) * 100
dd["hist-impl"] = (dd["sigma_hist"] - dd["sigma_impl"]) * 100

display(dd.groupby("contract")[["C-impl", "P-C", "hist-impl"]].agg(["mean", "median", "std"]))
print("C-impl：Call 腿位置的隐含波动率减 σ_impl；P-C：Put 腿减 Call 腿；hist-impl：历史波动率减 σ_impl")

atm = dd[dd["contract"] == "atm"].assign(year=lambda x: x["date"].dt.year)
by_year = atm.groupby("year")["P-C"].agg(["mean", "median", "min", "max"])
by_year["P-C>0 的天数占比"] = atm.groupby("year")["P-C"].agg(lambda s: (s > 0).mean())
display(by_year.round(2))
```

**曲面与两条腿.** 校准得到的 Heston 隐含波动率曲面具有负偏斜（$\rho$ 在 $-0.69$ 至 $-0.19$），微笑的曲率随期限衰减。chooser 的两条腿位于曲面上同一远期价内外程度，$\ln(K/F(T_2))=\ln(K'/F(T_1))=\ln(K/S)-(r-q)T_2$，只有期限不同，所以两条腿隐含波动率之差只反映该位置的期限结构。在 2021-12-01、2024-06-03 与各参数中位数的快照上，$T_1$ 与 $T_2$ 的平值隐含波动率之差（Put 腿减 Call 腿）为 $-0.52$、$-0.80$ 与 $-1.14$ 个波动率点，期限结构向上；在期限结构倒挂的 2020-04-01 为 $+4.89$。BSM 以单一波动率为两条腿定价，期限结构向上时 Put 腿偏高，倒挂时偏低，这与 $K=S_t$ 合约 Put 腿的平均偏差为正（$+0.30$）的符号相符。

### 4 情绪缺口

```python
# 误差与情绪的相关性，只是相关，两个模型都不含情绪
s = df.dropna(subset=["sentiment_score", "vix", "bsm_hist", BENCH]).copy()
s["rel_err"] = (s["bsm_hist"] - s[BENCH]) / s[BENCH] * 100
s["abs_rel_err"] = s["rel_err"].abs()
print(f"有情绪分的交易日占比：{df['sentiment_score'].notna().mean():.1%}")


def rank_corr(x, y):
    return np.corrcoef(rankdata(x), rankdata(y))[0, 1]


def partial_rank_corr(x, y, z):
    """秩的偏相关：把 x、y 的秩对 z 的秩回归，取残差再算相关"""
    rx, ry, rz = rankdata(x), rankdata(y), rankdata(z)
    resid = lambda a: a - np.polyval(np.polyfit(rz, a, 1), rz)
    return np.corrcoef(resid(rx), resid(ry))[0, 1]


rows = []
for contract, g in s.groupby("contract"):
    for target in ["rel_err", "abs_rel_err"]:
        for var in ["sentiment_score", "news_count"]:
            gg = g.dropna(subset=[var])
            rows.append({"合约": contract, "误差": target, "变量": var, "天数": len(gg),
                         "秩相关": rank_corr(gg[var], gg[target]),
                         "控制 VIX 后的偏相关": partial_rank_corr(gg[var], gg[target], gg["vix"])})
display(pd.DataFrame(rows).set_index(["合约", "误差", "变量"]))

s["情绪分组"] = s.groupby("contract")["sentiment_score"].transform(lambda x: pd.qcut(x, 3, labels=["低", "中", "高"]))
display(s.groupby(["contract", "情绪分组"], observed=True)
        .agg(天数=("rel_err", "size"), 平均相对偏差_pct=("rel_err", "mean"), 平均绝对相对偏差_pct=("abs_rel_err", "mean")))
```

**情绪缺口.** 在 1531 个有情绪分的交易日上，误差与情绪分的秩相关在 $-0.05$ 到 $+0.02$ 之间；控制 VIX 后，情绪分与有符号误差的偏相关为 $-0.01$（$K=S_t$）与 $-0.04$（$K=150$），情绪分在 VIX 之外对误差没有解释力。与新闻篇数的偏相关分别为 $-0.13$ 与 $-0.24$（有符号误差）、$-0.01$ 与 $+0.14$（绝对误差），幅度较小。按情绪分三等分，$K=S_t$ 的平均相对偏差由 $-4.0\%$ 经 $-6.7\%$ 到 $-8.6\%$，平均绝对相对偏差由 27.6% 降至 25.4%，几乎不随情绪变化。以上均为相关性，两个模型都不含情绪；日度序列自相关强，有效样本远少于 1531 天，未给出显著性检验。

### 5 总结

本节以与市场期权价格一致的 Heston 模型为基准，检验常数波动率的 BSM 对 chooser 期权的定价。基准的蒙特卡罗误差（标准误约 0.07，离散偏差约 0.005）远小于其校准误差（折算为价格约 0.4–0.6），后者决定了差异能否被解读为模型差异。BSM 的表现取决于波动率输入：以历史波动率为输入时 RMSE 为 9.44 与 7.33，偏差随 VIX 变号，2020 年 3–4 月达到 35.6；以 Heston 隐含波动率为输入时降至 0.50 与 1.34。总误差的主要部分来自波动率估计，常数波动率假设带来的曲面形状误差较小：$K=S_t$ 合约的绝对差中位数（0.36）小于折算的校准不确定性（0.58），仅 30% 的交易日超过它，无法与之区分；$K=150$ 合约的（0.95）大于校准不确定性（0.39），81% 的交易日超过它，可以辨别。误差与情绪分在控制 VIX 后没有相关。上述比较的是 BSM 与 Heston 的模型差异，不是对市场价格的误差，因为 chooser 没有市场成交价。
