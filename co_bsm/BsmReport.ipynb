### Chooser 期权的定价分解

设当前时刻为 $0$，股价为 $S_0$。Chooser 期权赋予持有者在选择日 $T_1$ 决定其为看涨或看跌期权的权利，二者行权价均为 $K$、到期日均为 $T_2 > T_1$。记 $\tau = T_2 - T_1$，无风险利率 $r$ 与连续股息率 $q$ 为常数，市场无套利。记 $C_t$、$P_t$ 为 $t$ 时刻行权价 $K$、$T_2$ 到期的欧式看涨、看跌期权价值，$(y)^+ = \max(y, 0)$。

理性持有者在 $T_1$ 选择价值较高者，故 chooser 期权在 $T_1$ 的价值为

$$
V_{T_1} = \max\left(C_{T_1},\ P_{T_1}\right).
$$

利用恒等式 $\max(a, b) = a + (b - a)^+$，可将其写为

$$
V_{T_1} = C_{T_1} + \left(P_{T_1} - C_{T_1}\right)^+ .
$$

由看跌-看涨平价，在 $T_1$ 时刻有

$$
P_{T_1} - C_{T_1} = K e^{-r\tau} - S_{T_1} e^{-q\tau},
$$

其右端仅依赖于 $S_{T_1}$，不含期权价值。代入并提出正常数 $e^{-q\tau}$，得

$$
V_{T_1} = C_{T_1} + e^{-q\tau}\left(K' - S_{T_1}\right)^+, \qquad K' = K e^{-(r-q)\tau}.
$$

上式第二项恰为 $e^{-q\tau}$ 份行权价为 $K'$、于 $T_1$ 到期的欧式看跌期权的到期价值。

在风险中性测度 $\mathbb{Q}$ 下，chooser 期权的当前价值为其 $T_1$ 价值的折现期望：

$$
V_0 = e^{-rT_1}\,\mathbb{E}^{\mathbb{Q}}\left[C_{T_1}\right] + e^{-q\tau}\, e^{-rT_1}\,\mathbb{E}^{\mathbb{Q}}\left[\left(K' - S_{T_1}\right)^+\right].
$$

对第一项，由 $C_{T_1} = e^{-r\tau}\,\mathbb{E}^{\mathbb{Q}}\left[(S_{T_2} - K)^+ \mid \mathcal{F}_{T_1}\right]$ 及全期望公式，

$$
e^{-rT_1}\,\mathbb{E}^{\mathbb{Q}}\left[C_{T_1}\right] = e^{-rT_2}\,\mathbb{E}^{\mathbb{Q}}\left[(S_{T_2} - K)^+\right] = \mathrm{Call}(K,\ T_2).
$$

第二项中的折现期望即为行权价 $K'$、到期日 $T_1$ 的欧式看跌期权的当前价格 $\mathrm{Put}(K',\ T_1)$。于是

$$
V_0 = \mathrm{Call}(K,\ T_2) + e^{-q\tau}\,\mathrm{Put}(K',\ T_1), \qquad K' = K e^{-(r-q)\tau}.
$$

该分解表明，chooser 期权等价于一份行权价 $K$、到期日 $T_2$ 的欧式看涨期权多头，与 $e^{-q\tau}$ 份行权价 $K'$、到期日 $T_1$ 的欧式看跌期权多头之和。后者的价值即为在 $T_1$ 将看涨期权转换为看跌期权这一权利的价值。相应地，$T_1$ 时刻的最优选择规则为

$$
P_{T_1} > C_{T_1} \iff S_{T_1} < K' \iff S_{T_1} e^{(r-q)\tau} < K,
$$

即当 $T_1$ 时刻的远期价格低于 $K$ 时选择看跌期权。

若在恒等式中以 $P_{T_1}$ 为基准，即 $\max(a, b) = b + (a - b)^+$，同理可得对称形式

$$
V_0 = \mathrm{Put}(K,\ T_2) + e^{-q\tau}\,\mathrm{Call}(K',\ T_1).
$$

考察两种极限情形。当 $T_1 = T_2$ 时，$\tau = 0$，$K' = K$，分解退化为 $V_0 = \mathrm{Call}(K,\ T_2) + \mathrm{Put}(K,\ T_2)$，即跨式组合；当 $T_1 \to 0$ 时，看跌期权部分收敛于 $(P_0 - C_0)^+$，从而 $V_0 \to \max(C_0,\ P_0)$。因此 chooser 期权的价值随 $T_1$ 在 $\max(C_0, P_0)$ 与跨式组合价值之间变化。

上述推导仅依赖看跌-看涨平价、无套利及 $r$、$q$ 为常数的假设，未对股价分布作任何设定，故分解在 Black-Scholes-Merton 模型与随机波动率模型下均成立；具体模型仅用于确定 $\mathrm{Call}(K,\ T_2)$ 与 $\mathrm{Put}(K',\ T_1)$ 的价格。

### 平价证明 补充

设无风险利率 $r$ 与连续股息率 $q$ 为常数，市场无套利且不存在交易成本与卖空限制。考虑同一标的、行权价均为 $K$、到期日均为 $T_2$ 的欧式看涨与看跌期权，记其在 $t \in [0, T_2]$ 时刻的价值为 $C_t$ 与 $P_t$，剩余期限 $\tau_t = T_2 - t$。下面证明

$$
C_t - P_t = S_t e^{-q\tau_t} - K e^{-r\tau_t}.
$$

在 $t$ 时刻构造两个组合。组合 $\mathcal{A}$ 为多头一份看涨期权、空头一份看跌期权，其价值为 $C_t - P_t$。组合 $\mathcal{B}$ 为持有 $e^{-q\tau_t}$ 股股票并将股息即时全部再投资于该股票，同时以无风险利率借入 $K e^{-r\tau_t}$，其价值为 $S_t e^{-q\tau_t} - K e^{-r\tau_t}$。

组合 $\mathcal{A}$ 在到期日的价值为

$$
(S_{T_2} - K)^+ - (K - S_{T_2})^+ .
$$

当 $S_{T_2} \ge K$ 时，上式等于 $(S_{T_2} - K) - 0$；当 $S_{T_2} < K$ 时，上式等于 $0 - (K - S_{T_2})$。故对任意 $S_{T_2}$ 均有

$$
(S_{T_2} - K)^+ - (K - S_{T_2})^+ = S_{T_2} - K .
$$

对于组合 $\mathcal{B}$，记 $s \in [t, T_2]$ 时刻的持股数为 $n_s$。在 $[s, s + \mathrm{d}s]$ 内持仓获得股息 $q S_s n_s\,\mathrm{d}s$，以价格 $S_s$ 全部再投资，持股数增加 $q n_s\,\mathrm{d}s$，即

$$
\mathrm{d}n_s = q\, n_s\,\mathrm{d}s, \qquad n_t = e^{-q\tau_t},
$$

解得 $n_{T_2} = n_t e^{q\tau_t} = 1$。借款在到期日的本息合计为 $K e^{-r\tau_t} e^{r\tau_t} = K$。因此组合 $\mathcal{B}$ 在到期日的价值为 $S_{T_2} - K$，且建立后无需追加或抽回资金，为自融资组合。

两组合在到期日的价值对任意 $S_{T_2}$ 均相等。若二者在 $t$ 时刻价值不等，则买入价低者、卖出价高者，可在 $t$ 时刻获得严格为正的现金流，而到期日头寸恰好抵消，构成套利，与无套利假设矛盾。故二者在 $t$ 时刻价值相等，即

$$
C_t - P_t = S_t e^{-q\tau_t} - K e^{-r\tau_t}.
$$

上述证明仅依赖无套利与静态复制，未对 $S_{T_2}$ 的分布作任何假设。其原因在于组合 $\mathcal{A}$ 的到期价值是 $S_{T_2}$ 的线性函数，可由标的资产与无风险资产静态复制。等价地，在风险中性测度 $\mathbb{Q}$ 下，含股息再投资的折现股价为鞅，有 $\mathbb{E}^{\mathbb{Q}}[S_{T_2} \mid \mathcal{F}_t] = S_t e^{(r-q)\tau_t}$，由期望的线性性

$$
C_t - P_t = e^{-r\tau_t}\,\mathbb{E}^{\mathbb{Q}}\left[S_{T_2} - K \mid \mathcal{F}_t\right] = S_t e^{-q\tau_t} - K e^{-r\tau_t},
$$

结论相同。

取 $t = T_1$，记 $\tau = T_2 - T_1$，得

$$
P_{T_1} - C_{T_1} = K e^{-r\tau} - S_{T_1} e^{-q\tau},
$$

其右端仅依赖于 $T_1$ 时刻已知的股价 $S_{T_1}$，这一关系用于 chooser 期权的分解。

```python
def simulate_two_period(S, r, q, sigma, T1, T2, n_paths, seed=None):
    rng = np.random.default_rng(seed)
    z1 = rng.standard_normal(n_paths)
    z2 = rng.standard_normal(n_paths)
    tau = T2 - T1
    drift = r - q - 0.5 * sigma**2
    S_T1 = S * np.exp(drift * T1 + sigma * np.sqrt(T1) * z1)
    S_T2 = S_T1 * np.exp(drift * tau + sigma * np.sqrt(tau) * z2)
    return S_T1, S_T2


def mc_estimate(x):
    return x.mean(), x.std(ddof=1) / np.sqrt(len(x))


def chooser_payoffs(S_T1, S_T2, K, r, q, T1, T2, method):
    """
    返回每条路径折现到 0 时刻的 payoff。

    method = "path"       ：路径上直接选择，S_T1 >= K' 选 Call，否则选 Put
    method = "decomposed" ：分解形式，Call(K, T2) 与 e^{-q tau} Put(K', T1) 的 payoff 之和
    """
    tau = T2 - T1
    K_prime = K * np.exp(-(r - q) * tau)
    call_T2 = np.maximum(S_T2 - K, 0.0)
    put_T2 = np.maximum(K - S_T2, 0.0)

    if method == "path":
        return np.exp(-r * T2) * np.where(S_T1 >= K_prime, call_T2, put_T2)
    if method == "decomposed":
        return (np.exp(-r * T2) * call_T2
                + np.exp(-q * tau) * np.exp(-r * T1) * np.maximum(K_prime - S_T1, 0.0))
    raise ValueError(f"未知的 method：{method}")
```

```python
N_PATHS = 1_000_000
SEED = 20260923

S_T1, S_T2 = simulate_two_period(PAPER["S"], PAPER["r"], PAPER["q"], PAPER["sigma"],
                                 PAPER["T1"], PAPER["T2"], N_PATHS, seed=SEED)

args = (PAPER["K"], PAPER["r"], PAPER["q"], PAPER["T1"], PAPER["T2"])
labels = {"path": "路径上直接选择（S_T1 ≥ K'）", "decomposed": "分解形式"}
payoffs = {m: chooser_payoffs(S_T1, S_T2, *args, method=m) for m in labels}

rows = []
for m, x in payoffs.items():
    price, se = mc_estimate(x)
    rows.append({"方法": labels[m], "MC 价格": price, "标准误": se,
                 "与闭式解之差": price - chooser, "差 / 标准误": (price - chooser) / se})
mc_table = pd.DataFrame(rows)

K_prime = PAPER["K"] * np.exp(-(PAPER["r"] - PAPER["q"]) * tau)
print(f"闭式解：{chooser:.4f}    路径数：{N_PATHS:,}    K' = {K_prime:.4f}")
print(f"T1 时选择 Call 的路径占比：{(S_T1 >= K_prime).mean():.2%}\n")
print(mc_table.to_string(index=False))
```

```python
# 收敛性：价格估计随路径数的变化（95% 置信区间）
n_grid = np.unique(np.logspace(2, np.log10(N_PATHS), 30).astype(int))
fig, ax = plt.subplots(figsize=(10, 5.5))
for m, color in [("path", "C1"), ("decomposed", "C0")]:
    x = payoffs[m]
    est, se = np.array([mc_estimate(x[:n]) for n in n_grid]).T
    ax.plot(n_grid, est, color=color, label=labels[m])
    ax.fill_between(n_grid, est - 1.96 * se, est + 1.96 * se, color=color, alpha=0.2)
ax.axhline(chooser, color="k", ls="--", label=f"闭式解 {chooser:.2f}")
ax.set_xscale("log")
ax.set_xlabel("路径数 log")
ax.set_ylabel("价格")
ax.set_title("蒙特卡罗估计的收敛（阴影为 95% 置信区间）")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

蒙特卡罗模拟在风险中性测度下按几何布朗运动精确生成 $S_{T_1}$ 与 $S_{T_2}$，$T_1$ 时刻按 $S_{T_1} \ge K'$ 选择看涨期权。100 万条路径下，路径上直接选择与分解形式两种估计分别为 29.1247 与 29.1292，与闭式解之差分别为 0.17 与 0.03 个标准误，二者均与闭式解一致。分解形式的标准误为 0.0289，低于直接选择的 0.0306，方差约降低 11%。$T_1$ 时选择看涨期权的路径占比为 50.40%，与对数正态分布下 $P(S_{T_1} \ge K')$ 的理论值一致。

上述结果表明，闭式解与蒙特卡罗模拟相互印证，本 notebook 实现的 BSM chooser 期权定价模型是正确的，可作为后续与 Heston 随机波动率模型比较的基准。

### 参数敏感性分析
针对参数敏感性分析,Greeks的本质就是求偏导,由于求导是线性的运算符所以Chooser option对于Greeks的敏感性就可以通过分解公式来得出，我们给出如下的推导:

记当前时刻为 $t$，选择日与到期日为固定的日历时刻 $T_1 < T_2$，$\tau = T_2 - T_1$。无风险利率 $r$、连续股息率 $q$、波动率 $\sigma$ 为常数。记 $C(S, K, T; r, q, \sigma)$ 与 $P(S, K, T; r, q, \sigma)$ 为剩余期限 $T$ 的欧式看涨、看跌期权在 Black-Scholes-Merton 模型下的价格。由定价分解，chooser 期权在 $t < T_1$ 时的价值为

$$
V = C\left(S, K, T_2 - t\right) + w\, P\left(S, K', T_1 - t\right),
\qquad w = e^{-q\tau},\quad K' = K e^{-(r-q)\tau}.
$$

其中 $w$ 与 $K'$ 仅依赖于 $r$、$q$ 与 $\tau$，与 $S$、$\sigma$ 及当前时刻 $t$ 无关。下文以下标 $C$、$P$ 分别表示看涨腿 $C(S, K, T_2 - t)$ 与看跌腿 $P(S, K', T_1 - t)$ 的偏导数，$d_1'$、$d_2'$ 表示看跌腿的 $d_1$、$d_2$，$\phi$ 与 $N$ 分别为标准正态分布的密度与分布函数。

**对标的价格与波动率的偏导数.** 由于 $w$、$K'$ 与 $S$、$\sigma$ 无关，求导的线性性直接给出

$$
\Delta = \frac{\partial V}{\partial S} = \Delta_C + w\,\Delta_P,\qquad
\Gamma = \frac{\partial^2 V}{\partial S^2} = \Gamma_C + w\,\Gamma_P,\qquad
\mathcal{V} = \frac{\partial V}{\partial \sigma} = \mathcal{V}_C + w\,\mathcal{V}_P .
$$

**对时间的偏导数.** $V$ 通过两条腿的剩余期限 $T_2 - t$ 与 $T_1 - t$ 依赖于 $t$。二者之差恒为 $\tau$，故 $w$ 与 $K'$ 不随 $t$ 变化，从而

$$
\Theta = \frac{\partial V}{\partial t} = \Theta_C + w\,\Theta_P ,
$$

其中 $\Theta_C$ 按剩余期限 $T_2 - t$ 计算，$\Theta_P$ 按剩余期限 $T_1 - t$ 计算。

**看跌期权价格对行权价的偏导数.** 对 $P = K e^{-rT}N(-d_2) - S e^{-qT}N(-d_1)$ 关于 $K$ 求导，并注意 $\partial d_1/\partial K = \partial d_2/\partial K = -1/(K\sigma\sqrt{T})$，得

$$
\frac{\partial P}{\partial K}
= e^{-rT}N(-d_2) + \frac{1}{K\sigma\sqrt{T}}\left[K e^{-rT}\phi(d_2) - S e^{-qT}\phi(d_1)\right].
$$

由 $d_2 = d_1 - \sigma\sqrt{T}$ 可得 $\phi(d_2) = \phi(d_1)\exp\left(d_1\sigma\sqrt{T} - \tfrac{1}{2}\sigma^2 T\right) = \phi(d_1)\,\dfrac{S}{K}e^{(r-q)T}$，即 $K e^{-rT}\phi(d_2) = S e^{-qT}\phi(d_1)$，方括号内的项为零。于是

$$
\frac{\partial P}{\partial K} = e^{-rT}N(-d_2),
$$

对看跌腿即 $\partial P/\partial K' = e^{-r(T_1 - t)}N(-d_2')$。

**对无风险利率的偏导数.** $r$ 同时出现在两条腿的价格与 $K'$ 中，且 $\partial K'/\partial r = -\tau K'$。由链式法则，

$$
\rho = \frac{\partial V}{\partial r}
= \rho_C + w\left[\rho_P + \frac{\partial P}{\partial K'}\frac{\partial K'}{\partial r}\right]
= \rho_C + w\left[\rho_P - \tau K' e^{-r(T_1 - t)}N(-d_2')\right].
$$

**对股息率的偏导数.** $q$ 出现在两条腿的价格、$K'$ 与 $w$ 中，且 $\partial w/\partial q = -\tau w$，$\partial K'/\partial q = \tau K'$。由乘积法则与链式法则，

$$
\psi = \frac{\partial V}{\partial q}
= \psi_C + \frac{\partial w}{\partial q}P + w\left[\psi_P + \frac{\partial P}{\partial K'}\frac{\partial K'}{\partial q}\right]
= \psi_C - \tau w P + w\left[\psi_P + \tau K' e^{-r(T_1 - t)}N(-d_2')\right].
$$

**两条腿的 Greeks.** 上式中两条腿的各项偏导数由 Black-Scholes-Merton 公式给出。对剩余期限 $T$、行权价 $K$ 的欧式期权，记 $d_1 = \dfrac{\ln(S/K) + (r - q + \sigma^2/2)T}{\sigma\sqrt{T}}$，$d_2 = d_1 - \sigma\sqrt{T}$，有

$$
\begin{aligned}
\Delta_{\text{call}} &= e^{-qT}N(d_1), & \Delta_{\text{put}} &= -e^{-qT}N(-d_1),\\
\Gamma &= \frac{e^{-qT}\phi(d_1)}{S\sigma\sqrt{T}}, & \mathcal{V} &= S e^{-qT}\phi(d_1)\sqrt{T},\\
\Theta_{\text{call}} &= -\frac{S e^{-qT}\phi(d_1)\sigma}{2\sqrt{T}} - rK e^{-rT}N(d_2) + qS e^{-qT}N(d_1), &
\Theta_{\text{put}} &= -\frac{S e^{-qT}\phi(d_1)\sigma}{2\sqrt{T}} + rK e^{-rT}N(-d_2) - qS e^{-qT}N(-d_1),\\
\rho_{\text{call}} &= KT e^{-rT}N(d_2), & \rho_{\text{put}} &= -KT e^{-rT}N(-d_2),\\
\psi_{\text{call}} &= -ST e^{-qT}N(d_1), & \psi_{\text{put}} &= ST e^{-qT}N(-d_1).
\end{aligned}
$$

**适用范围.** 上述推导中，$\Delta$、$\Gamma$、$\mathcal{V}$、$\Theta$ 的可加性与 $\rho$、$\psi$ 的链式法则附加项仅依赖于分解公式本身及 $w$、$K'$ 的形式，与标的价格的分布无关；Black-Scholes-Merton 模型仅用于给出两条腿各自的偏导数。在随机波动率等其他模型下，只需将两条腿的偏导数替换为相应模型下的值，其中 $\partial P/\partial K'$ 取该模型下看跌期权价格对行权价的偏导数。

```python
from scipy.stats import norm


def bsm_greeks(S, K, r, T, sigma, q, kind):
    """欧式期权的解析 Greeks"""
    sqT = np.sqrt(T)
    d1 = (np.log(S / K) + (r - q + 0.5 * sigma**2) * T) / (sigma * sqT)
    d2 = d1 - sigma * sqT
    dq, dr = np.exp(-q * T), np.exp(-r * T)
    gamma = dq * norm.pdf(d1) / (S * sigma * sqT)
    vega = S * dq * norm.pdf(d1) * sqT
    common = -S * dq * norm.pdf(d1) * sigma / (2 * sqT)
    if kind == "call":
        return dict(
            delta=dq * norm.cdf(d1), gamma=gamma, vega=vega,
            theta=common - r * K * dr * norm.cdf(d2) + q * S * dq * norm.cdf(d1),
            rho=K * T * dr * norm.cdf(d2), psi=-S * T * dq * norm.cdf(d1),
            dK=-dr * norm.cdf(d2),
        )
    return dict(
        delta=-dq * norm.cdf(-d1), gamma=gamma, vega=vega,
        theta=common + r * K * dr * norm.cdf(-d2) - q * S * dq * norm.cdf(-d1),
        rho=-K * T * dr * norm.cdf(-d2), psi=S * T * dq * norm.cdf(-d1),
        dK=dr * norm.cdf(-d2),
    )


def chooser_greeks(S, K, r, sigma, q, T1, T2):
    """
        chooser
    """
    tau = T2 - T1
    w = np.exp(-q * tau)
    Kp = K * np.exp(-(r - q) * tau)
    c = bsm_greeks(S, K, r, T2, sigma, q, "call")
    p = bsm_greeks(S, Kp, r, T1, sigma, q, "put")
    put_val = bsm_put(BSMParams(S, Kp, r, T1, sigma, q))
    return dict(
        delta=c["delta"] + w * p["delta"],
        gamma=c["gamma"] + w * p["gamma"],
        vega=c["vega"] + w * p["vega"],
        theta=c["theta"] + w * p["theta"],
        rho=c["rho"] + w * (p["rho"] - tau * Kp * p["dK"]),
        psi=c["psi"] + w * (p["psi"] + tau * Kp * p["dK"] - tau * put_val),
    )


def chooser_greeks_fd(S, K, r, sigma, q, T1, T2, h=1e-4):
    """
    中心差分
    """
    f = lambda **kw: chooser_price(ChooserParams(**{**dict(S=S, K=K, r=r, sigma=sigma, q=q, T1=T1, T2=T2), **kw}))
    dS = S * h
    return dict(
        delta=(f(S=S + dS) - f(S=S - dS)) / (2 * dS),
        gamma=(f(S=S + dS) - 2 * f() + f(S=S - dS)) / dS**2,
        vega=(f(sigma=sigma + h) - f(sigma=sigma - h)) / (2 * h),
        theta=(f(T1=T1 - h, T2=T2 - h) - f(T1=T1 + h, T2=T2 + h)) / (2 * h),
        rho=(f(r=r + h) - f(r=r - h)) / (2 * h),
        psi=(f(q=q + h) - f(q=q - h)) / (2 * h),
    )
```

```python
# Table 2 参数下的 基准Greeks
p0 = dict(S=PAPER["S"], K=PAPER["K"], r=PAPER["r"], sigma=PAPER["sigma"], q=PAPER["q"],
          T1=PAPER["T1"], T2=PAPER["T2"])
v0 = dict(S=PAPER["S"], K=PAPER["K"], r=PAPER["r"], T=PAPER["T2"], sigma=PAPER["sigma"], q=PAPER["q"])

ana = chooser_greeks(**p0)
fd = chooser_greeks_fd(**p0)
call_g = bsm_greeks(**v0, kind="call")
put_g = bsm_greeks(**v0, kind="put")

names = ["delta", "gamma", "vega", "theta", "rho", "psi"]
greeks_table = pd.DataFrame({
    "Chooser（解析）": [ana[g] for g in names],
    "Chooser（差分）": [fd[g] for g in names],
    "Call(K, T2)": [call_g[g] for g in names],
    "Put(K, T2)": [put_g[g] for g in names],
    "跨式组合": [call_g[g] + put_g[g] for g in names],
}, index=names)

# 换成常用单位：vega、rho、psi 为参数变动 1 个百分点，theta 为每日
units = pd.Series({"delta": 1, "gamma": 1, "vega": 0.01, "theta": 1 / 365, "rho": 0.01, "psi": 0.01})
greeks_table = greeks_table.mul(units, axis=0)
greeks_table.index = ["Delta", "Gamma", "Vega（σ +1%）", "Theta（每日）", "Rho（r +1%）", "Psi（q +1%）"]
print("解析与差分的最大相对差：",
      f"{(greeks_table['Chooser（解析）'] / greeks_table['Chooser（差分）'] - 1).abs().max():.1e}")
greeks_table
```

```python
# 情景分析：完全重估，与 Greeks 的二阶近似对比
base = chooser_price(ChooserParams(**p0))
scenarios = {
    "S -10%": dict(S=p0["S"] * 0.9),
    "S +10%": dict(S=p0["S"] * 1.1),
    "σ ×1.5": dict(sigma=p0["sigma"] * 1.5),
    "σ -5%": dict(sigma=p0["sigma"] - 0.05),
    "r +2%": dict(r=p0["r"] + 0.02),
    "q +1%": dict(q=p0["q"] + 0.01),
    "过去 30 天": dict(T1=p0["T1"] - 30 / 365, T2=p0["T2"] - 30 / 365),
}

rows = []
for label, bump in scenarios.items():
    new = chooser_price(ChooserParams(**{**p0, **bump}))
    dS = bump.get("S", p0["S"]) - p0["S"]
    dv = bump.get("sigma", p0["sigma"]) - p0["sigma"]
    dr = bump.get("r", p0["r"]) - p0["r"]
    dq = bump.get("q", p0["q"]) - p0["q"]
    dt = p0["T1"] - bump.get("T1", p0["T1"])
    approx = (ana["delta"] * dS + 0.5 * ana["gamma"] * dS**2 + ana["vega"] * dv
              + ana["rho"] * dr + ana["psi"] * dq + ana["theta"] * dt)
    rows.append({"情景": label, "重估价格": new, "价格变化": new - base, "Greeks 近似": approx})

scenario_table = pd.DataFrame(rows).set_index("情景")
scenario_table["近似误差"] = scenario_table["Greeks 近似"] - scenario_table["价格变化"]
print(f"基准价格：{base:.4f}")
scenario_table
```

```python
# 龙卷风图：各参数分别上下调整 10%，其余参数不变
bump = 0.10
labels = {"S": "标的价格 S", "K": "行权价 K", "sigma": "波动率 σ", "r": "无风险利率 r",
          "q": "股息率 q", "T1": "选择日 T1", "T2": "到期日 T2"}

base = chooser_price(ChooserParams(**p0))
rows = []
for name, label in labels.items():
    lo = chooser_price(ChooserParams(**{**p0, name: p0[name] * (1 - bump)}))
    hi = chooser_price(ChooserParams(**{**p0, name: p0[name] * (1 + bump)}))
    rows.append({"参数": label, "下调 10%": lo - base, "上调 10%": hi - base,
                 "区间": f"{p0[name] * (1 - bump):.4g} - {p0[name] * (1 + bump):.4g}"})

tornado = pd.DataFrame(rows).set_index("参数")
# S、K 附近价格呈 U 形，上下调整可能同向，所以用两侧变化的最大绝对值排序
tornado["影响幅度"] = tornado[["下调 10%", "上调 10%"]].abs().max(axis=1)
tornado = tornado.sort_values("影响幅度")

fig, ax = plt.subplots(figsize=(10, 5.5))
y = np.arange(len(tornado))
h = 0.38  
for col, dy, color in [("下调 10%", -h / 2, "C0"), ("上调 10%", h / 2, "C3")]:
    ax.barh(y + dy, tornado[col], height=h, color=color, label=col)
    for yi, v in zip(y + dy, tornado[col]):
        ax.annotate(f"{v:+.2f}", (v, yi), xytext=(4 if v >= 0 else -4, 0), textcoords="offset points",
                    va="center", ha="left" if v >= 0 else "right", fontsize=9)
ax.set_yticks(y, [f"{i}\n({r})" for i, r in zip(tornado.index, tornado["区间"])])
ax.axvline(0, color="k", lw=0.8)
ax.set_xlabel(f"价格变化（基准价格 {base:.2f}）")
ax.set_title("Chooser 价格的参数敏感性（各参数单独 ±10%）")
ax.legend(loc="lower right")
ax.grid(axis="x", alpha=0.3)
lim = tornado[["下调 10%", "上调 10%"]].abs().max().max() * 1.25
ax.set_xlim(-lim, lim)
plt.tight_layout()
plt.show()

tornado.iloc[::-1]
```

由参数敏感性分析可知，chooser 期权的价格主要由标的价格 $S$、波动率 $\sigma$ 与行权价 $K$ 决定。其中 $K$、$T_1$、$T_2$ 为合约条款，签约后保持不变，其敏感性仅作为产品设计的参考，不纳入风险因子分析；时间流逝的影响由 Theta 反映。市场风险因子包括 $S$、$\sigma$、$r$ 与 $q$，由龙卷风图与情景分析可知，价格对 $r$ 与 $q$ 的敏感度较低，因此 Greeks 的敏感性分析集中于 $S$ 与 $\sigma$ 两个维度：前者考察 Greeks 随标的价格的变化，后者考察 Greeks 随波动率的变化。

```python
# Greeks 随波动率的变化，对比三种标的价格
sig_grid = np.linspace(0.05, 0.8, 300)
s_list = [120.0, PAPER["S"], 190.0]

fig, axes = plt.subplots(2, 3, figsize=(15, 8.5))
for s in s_list:
    g = chooser_greeks(**{**p0, "S": s, "sigma": sig_grid})
    for ax, (key, title, u) in zip(axes.flat, panels):
        ax.plot(sig_grid * 100, g[key] * u, label=f"S = {s:g}")
for ax, (key, title, u) in zip(axes.flat, panels):
    ax.axvline(PAPER["sigma"] * 100, color="gray", ls=":", lw=1)
    ax.set(title=title, xlabel="σ (%)")
    ax.grid(alpha=0.3)
axes[0, 0].legend()
fig.suptitle("Chooser Greeks 随波动率的变化（K = 150，T1 = 0.5，T2 = 1）")
plt.tight_layout()
plt.show()

# 基准点的二阶 Greeks（差分）：Vanna = ∂Δ/∂σ，Volga = ∂Vega/∂σ
h = 1e-4
up = chooser_greeks(**{**p0, "sigma": p0["sigma"] + h})
dn = chooser_greeks(**{**p0, "sigma": p0["sigma"] - h})
pd.Series({
    "Vanna（σ +1% 时 Delta 的变化）": (up["delta"] - dn["delta"]) / (2 * h) * 0.01,
    "Volga（σ +1% 时 Vega 的变化，Vega 按 σ +1% 计）": (up["vega"] - dn["vega"]) / (2 * h) * 0.01 * 0.01,
}, name="Table 2 参数")
```

### 结论

在论文 Table 2 参数下，chooser 期权的 Greeks 与跨式组合相近：Delta 为 0.166，价格对标的价格呈 U 形，Vega 为 1.02（σ 每上升 1 个百分点），体现出做多波动率的性质。由于看跌腿于 $T_1$ 到期、期限短于看涨腿，其 Gamma（0.0208）高于同期限跨式组合（0.0172），每日 Theta（−0.054）更负，Vega 则略低于跨式组合（1.19）。Rho 与 Psi 分别为 −0.032 与 −0.260（参数每变动 1 个百分点），其计算需计入 $ K $ 与 $e^{-q\tau}$ 对 $r$、$q$ 的依赖，解析结果与数值差分的相对差在 $10^{-6}$ 量级。参数敏感性方面，各参数单独变动 ±10% 时，价格对标的价格、波动率与行权价最为敏感，标的价格与行权价的上下调整均使价格上升，对选择日与到期日次之，对股息率与无风险利率的敏感度较低。情景分析中，波动率放大 1.5 倍使价格由 29.13 升至 43.45，无风险利率上升 2 个百分点使价格仅变化 0.04，时间经过 30 天使价格下降 1.68；利率情景下 Greeks 的一阶近似与完全重估的变化方向相反，说明价格对利率的变化主要来自高阶效应。

Greeks 随标的价格的变化显示，Gamma 与 Theta 集中于行权价附近，选择日越近，二者在行权价附近的峰值越陡；Vega 随选择日推迟而增大。Greeks 随波动率的变化显示，在当前接近平值的位置，Vega 在较宽的波动率范围内近似恒定，Vanna 与 Volga 分别为 0.0007 与 −0.0003，价格对波动率的暴露近似线性；标的价格偏离行权价时，Vega 随波动率上升而增大，Volga 为正。波动率升高时，不同标的价格下的 Delta 向中间收敛，Theta 近似线性地变得更负。综合来看，chooser 期权的主要市场风险来自波动率与标的价格的凸性，时间衰减是持有该暴露的成本，利率与股息率风险较小。
